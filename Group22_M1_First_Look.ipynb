## 1. Problem statement and research questions

**Problem statement.** Violent and disruptive incidents are a daily concern in parts of Nigeria, but they are not spread evenly across states, months or types of event. This project uses ACLED's weekly state-level data for Nigeria, from January 2018 to October 2026, to describe where and when incidents occur and which kinds are most likely to be deadly. We then build simple classification models to predict whether a weekly state-level record includes at least one reported fatality. This is a sensitive topic: we report patterns factually and respectfully, and we do not assign blame to any person or group.

**Research questions**

| # | Research question | Columns used |
|---|---|---|
| RQ1 | How do the number of recorded events and the number of reported fatalities vary by year, month and state? | `week`, `admin1`, `events`, `fatalities` |
| RQ2 | Which states are the main hotspots for events and fatalities, and has this pattern changed since 2018? | `admin1`, `week`, `events`, `fatalities` |
| RQ3 | What share of weekly state-level records involve at least one fatality, and how does this differ by event type, sub-event type, state and season? | `event_type`, `sub_event_type`, `admin1`, `week`, `fatalities` |
| RQ4 | Which features (event type, state, month, number of events) are most associated with a record having at least one fatality? | all of the above (answered with models in M5) |

## 2. Data source and acquisition

**Source:** ACLED (Armed Conflict Location & Event Data), a conflict monitor that collects and publishes data on political violence and demonstrations (https://acleddata.com).
**Dataset:** ACLED aggregated data for Nigeria, downloaded as a CSV from ACLED's Data Export Tool with the "Aggregated" data type, using a registered myACLED account.
**Accessed (downloaded):** 13 November 2026
**File:** `ACLED_Data_2026-11-13_event_date_from_2018-01-01_event_date_to_2026-11-13.csv` (about 3.0 MB, 24,013 rows × 12 columns)
**Unit of analysis:** one row = one week (Saturday to Friday) × one state (Admin 1) × one sub-event type.
**Coverage returned:** weeks starting 30 December 2017 to 31 October 2026 (462 distinct weeks).

**Filters used in the Data Export Tool**

| Filter | Setting |
|---|---|
| Data type | Aggregated (by week and Admin 1) |
| Country | Nigeria (country code 566) |
| Event date | 1 January 2018 to 13 November 2026 |
| Event types selected (5) | Battles; Explosions/Remote violence; Protests; Riots; Violence against civilians |
| Sub-event types selected (16) | Armed clash; Attack; Shelling/artillery/missile attack; Air/drone strike; Mob violence; Violent demonstration; Remote explosive/landmine/IED; Looting/property destruction; Abduction/forced disappearance; Disrupted weapons use; Arrests; Grenade; Excessive force against protesters; Sexual violence; Suicide bomb; Chemical weapon |

ACLED's filters are additive, so selecting an event type returns all of its sub-event types (for example Peaceful protest), and the selected sub-event types Arrests, Disrupted weapons use and Looting/property destruction add three "Strategic developments" sub-event types. The file contains 6 event types and 19 sub-event types in total (no Chemical weapon rows were returned for Nigeria).

## 4. Data dictionary

Unit of analysis: one row = one week × one state (Admin 1) × one sub-event type. Types are as pandas loads the file. The same table is in `Group22_M1_Data_Dictionary.xlsx`.

| # | Column | Type | Description | Example |
|---|---|---|---|---|
| 1 | `week` | text (date) | Date of the Saturday that starts the week of aggregated data (Saturday to Friday). Needs converting to a date. | 2017-12-30 |
| 2 | `region` | text | ACLED region the location belongs to (Western Africa for Nigeria; International Waters for maritime rows). | Western Africa |
| 3 | `country` | text | Country of the events (Nigeria; Atlantic Ocean for 2 maritime rows). | Nigeria |
| 4 | `admin1` | text | First-level administrative area: a state or the Federal Capital Territory (plus one ocean area for the maritime rows). | Adamawa |
| 5 | `event_type` | text | ACLED event type: Battles, Explosions/Remote violence, Protests, Riots, Strategic developments, Violence against civilians. | Explosions/Remote violence |
| 6 | `sub_event_type` | text | Most granular ACLED event classification (19 values in this file). | Remote explosive/landmine/IED |
| 7 | `events` | integer | Number of discrete events recorded for that week, state and sub-event type. | 1 |
| 8 | `fatalities` | integer | Total reported fatalities across those events. | 3 |
| 9 | `population_exposure` | text (number with commas) | ACLED's best estimate of the population exposed to the events, based on proximity. Not to be summed across rows. Needs converting to a number. | 50,964 |
| 10 | `disorder_type` | text | ACLED's broad category: Political violence, Demonstrations, Strategic developments, or "Political violence; Demonstrations" where a sub-event type belongs to both. | Political violence |
| 11 | `centroid_latitude` | float | Latitude of the centre point of the state (for mapping), not of the event. | 9.170 |
| 12 | `centroid_longitude` | float | Longitude of the centre point of the state (for mapping), not of the event. | 12.295 |

Sources for the definitions: ACLED's aggregated-data guide (`week`, `sub_event_type`, `events`, `population_exposure`, centroids) and the ACLED Codebook with the values observed in this file (the other columns).

## 5. First look in pandas

The raw file is loaded exactly as downloaded. Nothing is changed in this notebook.

````python
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 140)

df = pd.read_csv("ACLED Data_2026-11-13_event_date_from_2018-01-01_event_date_to_2026-11-13.csv")   
````

````python
# Shape: (rows, columns)
df.shape
````

````python
# First five rows
df.head()
````

````python
# Column names, data types and non-null counts
df.info()
````

````python
# Missing values per column
pd.DataFrame({"missing": df.isna().sum(), "percent": (df.isna().mean() * 100).round(2)})
````

````python
# Duplicate rows
print("Fully duplicated rows:", df.duplicated().sum())
print("Repeated week + state + sub-event type combinations:", df.duplicated(["week", "admin1", "sub_event_type"]).sum())
````

````python
# Time coverage. "week" is the Saturday that starts each Saturday-to-Friday week.
weeks = pd.to_datetime(df["week"])
print("First week starts:", weeks.min().date())
print("Last week starts: ", weeks.max().date())
print("Distinct weeks:   ", weeks.nunique())

per_year = df.groupby(weeks.dt.year)[["events", "fatalities"]].sum()
per_year.insert(0, "rows", weeks.dt.year.value_counts().sort_index())
per_year.index.name = "year of week start"
per_year
````

````python
# Numeric columns
df[["events", "fatalities", "centroid_latitude", "centroid_longitude"]].describe().round(2)
````

````python
# Categorical columns: number of distinct values
for col in ["region", "country", "admin1", "event_type", "sub_event_type", "disorder_type"]:
    print(f"{col}: {df[col].nunique()} distinct values")
````

````python
# Event types and sub-event types: rows, events and fatalities
df.groupby(["event_type", "sub_event_type"]).agg(
    rows=("events", "size"), events=("events", "sum"), fatalities=("fatalities", "sum")
)
````

````python
# States (Admin 1): the ten with the most events
states = df.groupby("admin1").agg(events=("events", "sum"), fatalities=("fatalities", "sum"))
states.sort_values("events", ascending=False).head(10)
````

````python
# Rows that are not in Nigeria
df[df["country"] != "Nigeria"]
````

````python
# population_exposure is stored as text with thousands separators
print("dtype:", df["population_exposure"].dtype)
print("Rows with a comma in the value:", df["population_exposure"].str.contains(",").sum())
print("Rows equal to 0:", (df["population_exposure"] == "0").sum())
df["population_exposure"].head()
````

````python
# Preview of the planned target: does a row record at least one fatality?
has_fatality = df["fatalities"] >= 1
print("Rows with at least one fatality:", has_fatality.sum(), f"({has_fatality.mean():.1%})")

(df.assign(has_fatality=has_fatality)
   .groupby("event_type")["has_fatality"].mean().mul(100).round(1)
   .rename("% of rows with a fatality").to_frame())
````
